```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load("trajectory.npz")
xt = data["xt"]
yt = data["yt"]
inp = data["inp"]
t = data["t"]
plt.figure()
plt.plot(xt, yt)
plt.xlabel("x")
plt.ylabel("y")
plt.axis('equal')
plt.title("Trajectory")
plt.grid(True)
plt.show()
```

```python
plt.plot(t[900:], inp[900:])

plt.xlabel("input")
plt.ylabel("time")
plt.grid(True)
plt.show()
```
